```python
# IrriBot - Smart Irrigation Agent
# Predefined Rule-Based Agent
# Agent decides whether to turn the water pump ON or OFF
# based on soil moisture, temperature, humidity and rain condition

soil_moisture = int(input("Enter Soil Moisture Level (%): "))
temperature = int(input("Enter Temperature (Celsius): "))
humidity = int(input("Enter Humidity Level (%): "))
rain = input("Is it raining right now? (yes/no): ").lower()

print("\n--- IrriBot is analyzing the environment ---\n")

if rain == "yes":
    print("Rain detected.")
    print("Decision: Pump OFF")
    print("Reason: Natural rainfall is enough, no need to irrigate.")

elif soil_moisture < 30:
    print("Soil moisture is very low.")
    if temperature > 35:
        print("Decision: Pump ON (High intensity)")
        print("Reason: Dry soil and high temperature, plants need immediate water.")
    else:
        print("Decision: Pump ON (Normal intensity)")
        print("Reason: Soil moisture is below the safe threshold.")

elif 30 <= soil_moisture <= 60:
    if humidity < 40 and temperature > 30:
        print("Decision: Pump ON (Light watering)")
        print("Reason: Moderate soil moisture but hot and dry air increases water loss.")
    else:
        print("Decision: Pump OFF")
        print("Reason: Soil moisture is at an acceptable level.")

else:
    print("Soil moisture is sufficient.")
    print("Decision: Pump OFF")
    print("Reason: Watering right now would waste water and could harm the plants.")

print("\n--- IrriBot will continue monitoring conditions ---")
```
