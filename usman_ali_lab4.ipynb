```python
!pip install -U google-genai

from google import genai

client = genai.Client(api_key="")

MODEL = "models/gemini-3.5-flash"


def FUUAST_University(user_input):

    university_data = {
        "FUUAST": {
            "campuses": {

                "Karachi": {
                    "BSCS": {
                        "Fall 2024": {

                            "Database Systems": {
                                "teacher": "Miss. Uzma Khan"
                            },

                            "Data Structures": {
                                "teacher": "Dr. Shazia "
                            }

                        }
                    }
                },

                "Islamabad": {
                    "BSCS": {
                        "Fall 2024": {

                            "Database Systems": {
                                "teacher": "Dr. Sarim Khan"
                            },

                            "Data Structures": {
                                "teacher": "Dr. Uzma"
                            }

                        }
                    }
                }

            }
        }
    }


    prompt = f"""
    You are a professional FUUAST Univesity agent in Pakistan.

    Your job is to answer the users questions accurately.

    You have access to the following private university:

    University Information:
    {university_data}

    1. Use univeristy information to answer the users question.
    2. Do not invent university-specific information.
    3. If you didn't find information, use your own knowledge.
    4. If you don't know the answer, say that you don't know.

    Make response detailed but easy for students.

    User request: {user_input}

    Answer:
    """


    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("FUUAST University Agent Started (type 'exit' to stop)\n")


while True:

    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:

        reply = FUUAST_University(user)

        print("\nAgent:\n", reply, "\n")

    except Exception as e:

        print("Error:", e)
```
